```python
# Agregando duas listas com Zip

# Armazenar mais de uma informação em variáveis
# Manter a sequencia dos dados em uma variavel

produtos = ['arroz', 'feijão', 'macarrão', 'atum', 'azeite']

valores = [10, 20, 30, 40, 50]

duas_listas = zip(produtos, valores)

print(list(duas_listas))
```

```python
# Criando Sets

# Similar a listas
# Evita itens duplicados
# Não utiliza index

list1 = [10, 20, 30, 40, 50]
list2 = [10, 20, 60, 70, 80]

num1 = set(list1)
num2 = set(list2)

# resultado_ordenado = sorted(num1 | num2) #Union

# resultado_ordenado = sorted(num1 - num2) # Difference

# resultado_ordenado = sorted(num1 ^ num2) # Symmetric Difference

resultado_ordenado = sorted(num1 & num2) # Intersection

print(resultado_ordenado)
```

```python
# Funções em Sets

# Similar a listas
# Evita itens duplicados
# Não utiliza index

list1 = [10, 20, 30, 40, 50, 60]
s1 = {10, 20, 30, 40, 50, 60}
s1.add(70)
s1.update([80,90])

resultado_ordenado = sorted(s1)
print(resultado_ordenado)
```
