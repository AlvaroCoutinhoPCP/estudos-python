```python
# Looping dentro de uma lista

# Armazenar mais de uma informação em variáveis
# Manter a sequencia dos dados em uma variavel

valores = [50, 80, 110, 150, 170]

for x in valores:
    print(f'O valor final do produto é de R${x}')

#
```

```python
# Verificando itens em uma lista

# Armazenar mais de uma informação em variáveis
# Manter a sequencia dos dados em uma variavel

cor_cliente = input('Digite a cor desejada: ')
cores = ['amarelo', 'azul', 'vermelho']

if cor_cliente. lower() in cores:
    print('Em estoque')
else:
    print('Fazer pedido')

```

```python
# Agregando duas listas com Zip

# Armazenar mais de uma informação em variáveis
# Manter a sequencia dos dados em uma variavel

produtos = ['arroz', 'feijão', 'macarrão', 'atum', 'azeite']

valores = [10, 20, 30, 40, 50]

duas_listas = zip(produtos, valores)

print(list(duas_listas))
```
