```python
# De functions a Libraries

'''
# Hierarquia de Organização de Código em Python

*Function (Função):** Bloco de código reutilizável que executa uma tarefa específica e isolada;

*Module (Módulo):** Um único arquivo `.py` que agrupa funções,
variáveis e classes relacionadas a um mesmo propósito;

*Package (Pacote):** Uma pasta (diretório) que organiza múltiplos módulos estruturados
sob um mesmo tema ou funcionalidade;

*Library (Biblioteca):** Conjunto completo de pacotes e módulos
prontos para resolver uma grande área de problemas (ex: Pandas, NumPy).

'''

```

```python
# Como funciona uma função?

def boas_vindas():
  print('Olá!')
  print('Seja bem vindo!')

boas_vindas()
```

```python
# Argumentos default e non default

# Default = Aquele que voce define o valor no parametro
# Non_Default = Aquele que voce não define o valor no parametro

def boas_vindas(nome, quantidade=7):
  print(f'Olá {nome}!')
  print(f'Temos {quantidade} laptops em estoque')

boas_vindas('Marcos')
boas_vindas('Ronaldo')
boas_vindas('Linda')
```

```python
# Print ou retur em funções

def cliente1(nome):
  print(f'Olá {nome}')


cliente1('Maria')

def cliente2(nome):
  return f'Olá {nome}'

cliente2('José')

#

x = cliente1('Maria')
y = cliente2('José')

print('Valor de x:', x) # Vai imprimir: Valor de x: None (Vazio! O print não devolveu nada)
print('Valor de y:', y) # Vai imprimir: Valor de y: Olá José (O dado foi salvo!)

```

```python
# Varios argumentos, X*argas com números

def soma(*numeros):
  resultado = 0
  for num in numeros:
    resultado = resultado + num
  return resultado


x = soma(1,2,3,4,5,6,7,8,9)

print(x)
```

```python
# Varios argumentos, X*argas nomeando parametros

def agencia(**carro):
  return carro

print(agencia(marca='Gol', cor='Branco', motor=1.0, ano= 2017))
print(agencia(marca='Argo', cor='Prata', motor=1.0, ano= 2021))
print(agencia(marca='Versa', cor='Vermelho', motor=1.6, ano= 2024))
```
