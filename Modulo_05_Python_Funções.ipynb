```python
# De functions a Libraries

'''
# Hierarquia de Organização de Código em Python

*Function (Função):** Bloco de código reutilizável que executa uma tarefa específica e isolada;

*Module (Módulo):** Um único arquivo `.py` que agrupa funções,
variáveis e classes relacionadas a um mesmo propósito;

*Package (Pacote):** Uma pasta (diretório) que organiza múltiplos módulos estruturados
sob um mesmo tema ou funcionalidade;

*Library (Biblioteca):** Conjunto completo de pacotes e módulos
prontos para resolver uma grande área de problemas (ex: Pandas, NumPy).

'''

```

```python
# Como funciona uma função?

def boas_vindas():
  print('Olá!')
  print('Seja bem vindo!')

boas_vindas()
```
