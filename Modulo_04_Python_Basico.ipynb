```python
# Módulos, Bibliotecas & Stack de Dados em Python

'''
==============================================================================
1. MÉTODOS DE IMPORTAÇÃO DE BIBLIOTECAS
==============================================================================
- import biblioteca as apelido  -> Carrega a biblioteca inteira (ex: import pandas as pd)
- from biblioteca import recurso -> Carrega apenas uma função/classe específica (ex: from datetime import date)
- !pip install nome_biblioteca  -> Comando de terminal para instalar bibliotecas que não vêm pré-instaladas

==============================================================================
2. RESUMO DO STACK DE DADOS & CONCEITOS CHAVE
==============================================================================
- JSON (Módulo Nativo):
  Formato universal de texto (estruturado como dicionário) para troca de dados entre APIs/sistemas.
  * json.loads() -> Converte string JSON em Dicionário do Python.

- NumPy (Biblioteca Externa):
  O "motor" de computação numérica de alta performance.
  * Trabalha com ARRAYS: blocos contínuos e engessados (homogêneos) de dados na memória.
  * Por ter dados de um único tipo alocados lado a lado, permite cálculos matemáticos ultrarrápidos em lote.

- Pandas (Biblioteca Externa):
  O "Excelzão" em modo turbo do Python. Construído em cima do NumPy.
  * Trabalha com DATAFRAMES: tabelas estruturadas em linhas e colunas para ETL, limpeza e análise.

- Matplotlib (Biblioteca Externa):
  O motor de visualização de dados. Transforma tabelas e arrays em recursos visuais (gráficos).
'''

# ----------------------------------------------------------------------------
# EXEMPLOS PRÁTICOS INTEGRADOS
# ----------------------------------------------------------------------------

# 1. Imports
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import date

# --- Exemplo JSON (Leitura de texto vindo de API) ---
texto_json = '{"projeto": "Analise PCP", "status": "Ativo"}'
dados_api = json.loads(texto_json)
print(f"Projeto JSON carregado: {dados_api['projeto']}")

# --- Exemplo NumPy, Pandas, Datetime e Matplotlib trabalhando juntos ---

# NumPy: gerando array de dados numéricos (vendas aleatórias)
vendas_array = np.array([120, 250, 180, 310, 290])

# Pandas: estruturando o array em uma Tabela (DataFrame)
df = pd.DataFrame({
    'Mes': ['Jan', 'Fev', 'Mar', 'Abr', 'Mai'],
    'Vendas': vendas_array,
    'Ano_Registro': date.today().year  # Datetime nativo pegando o ano dinâmico
})

print("\n--- DataFrame Criado com Pandas ---")
print(df)

# Matplotlib: gerando um gráfico simples a partir dos dados do Pandas
plt.figure(figsize=(6, 3))
plt.bar(df['Mes'], df['Vendas'], color='skyblue')
plt.title(f'Vendas Mensais - {date.today().year}')
plt.xlabel('Mês')
plt.ylabel('Valor (R$)')
plt.show()
plt.show()
```

```python
# Exemplo Prático de PCP: NumPy + Pandas + Matplotlib (Contexto PCP madereira)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import date

# 1. NumPy: Vetores contínuos com a contagem de chapas por lote e espessura (mm)
chapas_por_lote = np.array([120, 150, 110, 140, 130]) # peças por lote
espessura_mm = np.array([15, 15, 18, 18, 25])         # espessura em mm

# 2. Pandas: Montagem do relatório da linha de produção (DataFrame)
df_producao = pd.DataFrame({
    'Lote': ['Lote A', 'Lote B', 'Lote C', 'Lote D', 'Lote E'],
    'Chapas_Produzidas': chapas_por_lote,
    'Espessura_mm': espessura_mm,
    'Data': date.today()  # Datetime nativo
})

# Calculando o volume aproximado em m³ usando operações vetoriais rápidas
# Fórmula simples: (Chapas * Espessura em metros * Fator de Área)
df_producao['Volume_m3'] = (df_producao['Chapas_Produzidas'] * (df_producao['Espessura_mm'] / 1000) * 5.7).round(2)

print("--- RELATÓRIO DE PRODUÇÃO (PCP) ---")
print(df_producao)
print(f"\nVolume Total Produzido: {df_producao['Volume_m3'].sum()} m³")

# 3. Matplotlib: Gráfico do volume produzido por lote
plt.figure(figsize=(7, 4))
plt.bar(df_producao['Lote'], df_producao['Volume_m3'], color='forestgreen')
plt.title(f'Volume de Produção por Lote (m³) - {date.today().strftime("%d/%m/%Y")}')
plt.xlabel('Lote de Produção')
plt.ylabel('Volume (m³)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
